```python
class Solution:
  def isPalindrome(self,x):
    if x < 0:
      return False
    org = x
    rev = 0
    while x > 0:
      d = x % 10
      rev = rev * 10 + d
      x //= 10
    return org == rev
```

```python
class Solution:
  def numOfSteps(self,num):
    steps = 0
    while num > 0:
     if num % 2 == 0:
        num = num // 2
     else:
        num = num - 1
     steps += 1

    return steps
```
