```python
class Solution:
  def isPalindrome(self,x):
    if x < 0:
      return False
    original = x
    reverse = 0
    while x > 0:
      digit = x % 10
      reverse = reverse * 10 + digit
      x //= 10
    return original == reverse
```

```python
class Solution:
  def numberOfSteps(self, num: int) -> int:
    steps = 0
    while num > 0:
     if num % 2 == 0:
        num = num // 2
     else:
        num = num - 1
     steps += 1

    return steps
```
